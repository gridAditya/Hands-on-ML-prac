```python
import numpy as np
import seaborn as sns

from matplotlib import pyplot as plt
```

## Simple Plots

```python
x = np.random.rand(100)
y = np.sin(x)

colors = np.random.rand(100)
```

```python
plt.figure(figsize=(6, 4))
scatter_obj = plt.scatter(x, y, alpha=0.5, c=colors, cmap="coolwarm", s=y * 100)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Customised Scatter Plot")
plt.colorbar(label="Color Values")
plt.grid(alpha=0.5)
plt.show()
```

## Line Plot

```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
plt.figure(figsize=(6, 4))
plt.plot(x, y1, c="red", label="Sine Function")
plt.plot(x, y2, c="green", label="Cosine Function")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Line Plots")
plt.grid(alpha=0.5)
plt.legend(loc="upper right")
plt.show()
```

## Histogram

```python
x_normal = np.random.normal(loc=0, scale=1, size=1000)
```

```python
plt.figure(figsize=(6,4))
plt.hist(x_normal, bins=25, color="green", edgecolor="black", alpha=0.8)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Histogram")
plt.grid(alpha=0.5)
plt.show()
```

## Bar Plots

```python
categories = ["rich", "middle", "poor"]
values = [100, 200, 300]
colors = ["red", "blue", "green"]

plt.figure(figsize=(6,5))
bars = plt.bar(categories, values, color=colors)
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Bar Plot")
plt.grid(alpha=0.5)

## Add Text to each bar
for bar, value in zip(bars, values):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, height + 1, f"{value}", ha="center", va="bottom") # va="bottom" places the bottom of text at the y-position

plt.show()
```

## Heatmaps

```python
x_heat = np.random.normal(loc=10, scale=1, size=(10,10))
```

```python
plt.figure(figsize=(7, 5))
sns.heatmap(x_heat, cmap="coolwarm", annot=True, fmt=".2f")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Heatmap")
plt.show()
```

## Writing Texts on Diagrams

```python
plt.text(0.4, 0.625, "Simple Text", color="black", fontsize=12, bbox=dict(facecolor="yellow", alpha=0.5, edgecolor="blue", boxstyle="round,pad=0.5"))
plt.text(0.4, 0.500, r'$f(x) = \sin(x)$', color="black", fontsize=12, bbox=dict(facecolor="yellow", alpha=0.5, edgecolor="blue", boxstyle="round,pad=0.5"))
plt.text(0.4, 0.375, r'$f(x)=\frac{1}{2}\cos(x)$', color="black", fontsize=12, bbox=dict(facecolor="yellow", alpha=0.5, edgecolor="blue", boxstyle="round,pad=0.5"))
plt.text(0.4, 0.250, r'$H_{2}O$', color="black", fontsize=12, bbox=dict(facecolor="yellow", alpha=0.5, edgecolor="blue", boxstyle="round,pad=0.5"))
plt.text(0.4, 0.125, r'$x^{10}$', color="black", fontsize=12, ha="center", va="bottom", bbox=dict(facecolor="yellow", alpha=0.5, edgecolor="blue", boxstyle="round,pad=0.5"))
plt.grid(alpha=0.5)
plt.show()
```
